```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

```python
cur.executescript('''
DROP TABLE IF EXISTS comida;

CREATE TABLE comida (
    id_platillo integer PRIMARY KEY,
    nombre varchar(50) NOT NULL,
    categoria varchar(30) NOT NULL,
    precio decimal(6,2) NOT NULL,
    disponible boolean default true
);

''')
conn.commit()
print("Tablas creadas: comida")
```

```python
# INSERT: alta de comida
run("""INSERT INTO comida (nombre, categoria, precio) VALUES
    ('coctel de camaron','ligero', 60.00),
    ('caldo de pollo', 'fuerte', 95.00),
    ('sopa fria', 'entrada', 30.00);""")

query("SELECT * FROM comida")
```

```python
# UPDATE: modificación de registros (aumento del precio de comidas fuertes)
run("UPDATE comida SET precio = precio * 1.10 WHERE categoria = 'fuerte';")
query("SELECT id_platillo, nombre, categoria, precio FROM comida WHERE categoria = 'fuerte'")
```

```python
# DELETE: eliminación de registros (se retira una comida)
run("DELETE FROM comida WHERE nombre = 'sopa fria';")
query("SELECT * FROM comida")
```
