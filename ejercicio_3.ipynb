```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

```python
cur.executescript('''
DROP TABLE IF EXISTS productos;

CREATE TABLE productos (
    id_productos integer PRIMARY KEY,
    nombre varchar(50) NOT NULL,
    tipo varchar(30) NOT NULL,
    precio decimal(6,2) NOT NULL,
    existencia integer
);

''')
conn.commit()
print("Tablas creadas: productos")
```

```python
# INSERT: alta de productos
run("""INSERT INTO productos (nombre, tipo, precio) VALUES
    ('jugos','beber', 20.00),
    ('chocolates', 'golosina', 13.00),
    ('sabritas', 'botana', 15.00);""")

query("SELECT * FROM productos")
```

```python
# UPDATE: modificación de registros (aumento del precio de productos golosina)
run("UPDATE productos SET precio = precio * 1.10 WHERE tipo = 'golosina';")
query("SELECT id_productos, nombre, tipo, precio FROM productos WHERE tipo = 'golosina'")
```

```python
# DELETE: eliminación de registros (se retira unos produtos)
run("DELETE FROM productos WHERE nombre = 'sabritas';")
query("SELECT * FROM productos")
```
